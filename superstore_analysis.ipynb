# 2. Statistical Hypothesis Testing

A significance level of **α = 0.05** is used for the statistical tests.

## Hypothesis 1 — Discount and Profit

### Research Question
Is there a statistically significant association between discount and profit?

**H₀:** There is no statistically significant association between discount and profit.

\[
H_0: \rho = 0
\]

**H₁:** There is a statistically significant association between discount and profit.

\[
H_1: \rho \neq 0
\]

Pearson and Spearman correlation tests are used.

> A significant correlation indicates association, not causation.

---

## Hypothesis 2 — Profitability Across Categories

### Research Question
Does average profitability differ across Furniture, Office Supplies, and Technology?

**H₀:** Mean profit is equal across all product categories.

\[
H_0: \mu_{Furniture}=\mu_{Office}=\mu_{Technology}
\]

**H₁:** At least one category has a different mean profit.

A one-way ANOVA and Kruskal-Wallis test are used. Tukey's HSD can be used for post-hoc pairwise comparisons when appropriate.

---

## Hypothesis 3 — Profitability Across Customer Segments

### Research Question
Does profitability differ across Consumer, Corporate, and Home Office segments?

**H₀:** Mean profitability is equal across all customer segments.

\[
H_0: \mu_{Consumer}=\mu_{Corporate}=\mu_{HomeOffice}
\]

**H₁:** At least one segment has a different mean profitability.

One-way ANOVA and Kruskal-Wallis tests are used.

---

## Hypothesis 4 — Profitability Across Regions

### Research Question
Does profitability differ across Central, East, South, and West?

**H₀:** Mean profitability is equal across all regions.

\[
H_0: \mu_{Central}=\mu_{East}=\mu_{South}=\mu_{West}
\]

**H₁:** At least one region has a different mean profitability.

One-way ANOVA and Kruskal-Wallis tests are used.

### Decision Rule

- **p-value < 0.05:** Reject H₀.
- **p-value ≥ 0.05:** Fail to reject H₀.

Statistical significance does not establish causation.

## 5. Outlier Analysis

Potential outliers are identified using the Interquartile Range (IQR) method.

\[
IQR = Q3-Q1
\]

\[
Lower\ Bound=Q1-1.5(IQR)
\]

\[
Upper\ Bound=Q3+1.5(IQR)
\]

Extreme observations are investigated rather than automatically removed because they may represent genuine high-value sales or significant losses.

```python
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1
outlier_summary = pd.DataFrame({'Q1': Q1,'Q3': Q3,'IQR': IQR,'Lower Bound': Q1 - 1.5 * IQR,'Upper Bound': Q3 + 1.5 * IQR})
outlier_summary
```

```python
outlier_counts = {}
for col in numeric_cols:
    lower = outlier_summary.loc[col, 'Lower Bound']
    upper = outlier_summary.loc[col, 'Upper Bound']
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_counts[col] = count
pd.Series(outlier_counts, name='Outlier Count')
```

```python
for col in numeric_cols:
    lower = outlier_summary.loc[col, 'Lower Bound']
    upper = outlier_summary.loc[col, 'Upper Bound']
    
    df[f'{col}_Outlier'] = (
        (df[col] < lower) | 
        (df[col] > upper)
    )
```

## 8. Category Profitability Analysis

Profitability is compared across Furniture, Office Supplies, and Technology.

The analysis uses descriptive statistics together with ANOVA and/or Kruskal-Wallis testing. When an overall significant difference is found, post-hoc comparisons can identify which groups differ.

```python
from scipy.stats import f_oneway

furniture_profit = df[
    df['Category'] == 'Furniture'
]['Profit']

office_profit = df[
    df['Category'] == 'Office Supplies'
]['Profit']

technology_profit = df[
    df['Category'] == 'Technology'
]['Profit']

f_stat, p_value = f_oneway(
    furniture_profit,
    office_profit,
    technology_profit
)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_result = pairwise_tukeyhsd(
    endog=df['Profit'],
    groups=df['Category'],
    alpha=0.05
)

print(tukey_result)
```

```python
from scipy.stats import kruskal

h_stat, p_value = kruskal(
    furniture_profit,
    office_profit,
    technology_profit
)

print("Kruskal-Wallis H-statistic:", h_stat)
print("p-value:", p_value)
```

```python
pairwise_tukeyhsd(
    endog=df['Profit'],
    groups=df['Category'],
    alpha=0.05
)
```

## 12. Loss-Making Product Analysis

Individual products and loss-making sub-categories are examined to understand how negative profitability is distributed.

A cumulative loss analysis is used to determine whether overall losses are concentrated among a relatively small number of products.

This helps identify areas that warrant further investigation into pricing, discounting, or cost structure.

```python
loss_products = loss_making_products[
    loss_making_products['Total_Profit'] < 0
].copy()

loss_products = loss_products.sort_values(
    'Total_Profit',
    ascending=True
)

loss_products['Cumulative_Loss'] = (
    -loss_products['Total_Profit']
).cumsum()

total_loss = -loss_products['Total_Profit'].sum()

loss_products['Cumulative_Loss_Pct'] = (
    loss_products['Cumulative_Loss'] / total_loss
) * 100

print("Total loss:", total_loss)

loss_products.head(15)
```
